```python
!pip install -q gradio sympy

import gradio as gr
import sympy as sp

def get_matrix(text):
    rows = text.strip().split("\n")
    return sp.Matrix([
        [sp.sympify(x) for x in row.replace(",", " ").split()]
        for row in rows
    ])

def calculate(matrix_text, operation):
    try:
        A = get_matrix(matrix_text)

        if operation == "Matrix":
            return str(A)

        if operation == "Transpose":
            return str(A.T)

        if operation == "Determinant":
            if A.rows != A.cols:
                return "Error: Matrix must be square."
            return str(A.det())

        if operation == "Inverse":
            if A.rows != A.cols:
                return "Error: Matrix must be square."
            if A.det() == 0:
                return "Inverse does not exist because determinant = 0."
            return str(A.inv())

        if operation == "Rank":
            return str(A.rank())

        if operation == "Eigenvalues":
            if A.rows != A.cols:
                return "Error: Matrix must be square."
            return str(A.eigenvals())

        if operation == "Eigenvectors":
            if A.rows != A.cols:
                return "Error: Matrix must be square."
            return str(A.eigenvects())

    except Exception as e:
        return "Error: " + str(e)


def solve_equations(text):
    try:
        equations = text.strip().split("\n")

        eqs = []
        variables = set()

        for line in equations:
            left, right = line.split("=")

            left = sp.sympify(left)
            right = sp.sympify(right)

            eqs.append(sp.Eq(left, right))
            variables.update(left.free_symbols)
            variables.update(right.free_symbols)

        variables = sorted(variables, key=str)

        solution = sp.solve(eqs, variables, dict=True)

        if not solution:
            return "No solution found."

        result = "LINEAR ALGEBRA SOLUTION\n"
        result += "========================\n\n"

        result += "Given Equations:\n"
        for eq in eqs:
            result += str(eq) + "\n"

        result += "\nSolution:\n"

        for sol in solution:
            for variable, value in sol.items():
                result += f"{variable} = {value}\n"

        return result

    except Exception as e:
        return "Error: " + str(e)


with gr.Blocks() as demo:

    gr.Markdown(
        """
        # 🧮 F4 - Linear Algebra Problem Solver

        Solve matrices and linear equations easily.
        """
    )

    with gr.Tab("Matrix Calculator"):

        matrix = gr.Textbox(
            label="Enter Matrix",
            value="2 3\n1 4",
            lines=5
        )

        operation = gr.Dropdown(
            [
                "Matrix",
                "Transpose",
                "Determinant",
                "Inverse",
                "Rank",
                "Eigenvalues",
                "Eigenvectors"
            ],
            value="Determinant",
            label="Select Operation"
        )

        button = gr.Button("Calculate")

        result = gr.Textbox(
            label="Result",
            lines=10
        )

        button.click(
            calculate,
            inputs=[matrix, operation],
            outputs=result
        )

    with gr.Tab("Linear Equation Solver"):

        equations = gr.Textbox(
            label="Enter Equations (one per line)",
            value="2*x+y=5\nx-y=1",
            lines=6
        )

        solve_button = gr.Button("Solve")

        solution = gr.Textbox(
            label="Solution",
            lines=15
        )

        solve_button.click(
            solve_equations,
            inputs=equations,
            outputs=solution
        )


demo.launch(share=True)
```
